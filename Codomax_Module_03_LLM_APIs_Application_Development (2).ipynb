```python

# Practical 1 – Managing Conversations

from google import genai
from google.colab import userdata

# Connect to Gemini
gemini_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=gemini_key)

# Create a conversation
chat = client.chats.create(
    model="gemini-3.5-flash-lite"
)

# First message
response1 = chat.send_message(
    "My name is Alex and I am learning Python."
)
print("Response 1:")
print(response1.text)

# Second message - uses previous conversation context
response2 = chat.send_message(
    "What programming language am I learning?"
)
print("\nResponse 2:")
print(response2.text)
```

```python

# Practical 2 – Understanding Message Roles

from google.genai import types

# Create a conversation with a system instruction
chat = client.chats.create(
    model="gemini-3.5-flash-lite",
    config=types.GenerateContentConfig(
        system_instruction="You are a helpful Python programming tutor. Keep answers simple."
    )
)

# User message
response1 = chat.send_message(
    "What is a Python variable?"
)

print("Assistant Response:")
print(response1.text)

# Another user message
response2 = chat.send_message(
    "Give me a simple example."
)

print("\nSecond Assistant Response:")
print(response2.text)

# Display conversation history
print("\nConversation History:")

for message in chat.get_history():
    print("Role:", message.role)
    print("Message:", message.parts[0].text)
    print("-" * 40)
```

```python

# Practical 3 – Model Parameters

from google.genai import types

prompt = "Explain Artificial Intelligence in simple terms."

# Configure model parameters
config = types.GenerateContentConfig(
    temperature=0.2,
    top_p=0.8,
    max_output_tokens=150
)

# Generate response
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=config
)

print("Generated Response:")
print(response.text)
```

```python

# Practical 4 – Token Usage Monitoring

prompt = """
Explain the importance of APIs in modern software development.
Include three benefits and one practical example.
"""

# Generate response
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=200
    )
)

# Display generated response
print("Generated Response:")
print(response.text)

# Inspect token usage
usage = response.usage_metadata

print("\nToken Usage:")
print("Input Tokens:", usage.prompt_token_count)
print("Output Tokens:", usage.candidates_token_count)
print("Total Tokens:", usage.total_token_count)
```

```python

# Restore Gemini API client

from google.colab import userdata
from google import genai

gemini_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=gemini_key)

print("Gemini client initialized successfully!")
```

```python

# Practical 1 – Generating Structured JSON Responses

from google.genai import types
import json

prompt = """
Provide information about Python in this structure:
{
    "language": "name",
    "type": "programming language type",
    "benefits": ["benefit 1", "benefit 2", "benefit 3"]
}
"""

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        temperature=0.2
    )
)

# Convert JSON response into a Python dictionary
data = json.loads(response.text)

print("Structured JSON Response:")
print(json.dumps(data, indent=4))
```

```python

# Practical 2 – Parsing and Validating JSON

import json

prompt = """
Return a JSON object containing:
name, language, and experience_level
for a fictional software developer.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        temperature=0.2
    )
)

try:
    data = json.loads(response.text)

    required_fields = [
        "name",
        "language",
        "experience_level"
    ]

    # Validate required fields
    missing_fields = [
        field for field in required_fields
        if field not in data
    ]

    if not missing_fields:
        print("JSON validation successful!")
        print("Developer:", data["name"])
        print("Language:", data["language"])
        print("Experience:", data["experience_level"])
    else:
        print("Missing fields:", missing_fields)

except json.JSONDecodeError:
    print("Invalid JSON response received.")
```

```python

# Practical 3 – Robust API Error Handling

import time

prompt = "Explain the importance of error handling in Python."

response = None

for attempt in range(1, 4):
    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        print("API request successful!")
        print(response.text)
        break

    except Exception as e:
        error_message = str(e)

        if "503" in error_message or "UNAVAILABLE" in error_message:
            print(f"Server unavailable. Attempt {attempt}/3")

        elif "429" in error_message or "RESOURCE_EXHAUSTED" in error_message:
            print("API rate limit or quota reached.")
            break

        else:
            print("API request failed:", type(e).__name__)
            break

        if attempt < 3:
            time.sleep(attempt * 5)

if response is None:
    print("Unable to complete the API request.")
```

```python

# Practical 4 – API Usage and Token Management

from google.genai import types

prompt = """
Explain three benefits of using APIs in software development.
Keep the explanation concise.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt,
    config=types.GenerateContentConfig(
        max_output_tokens=100,
        temperature=0.2
    )
)

print("Generated Response:")
print(response.text)

# Monitor token consumption
usage = response.usage_metadata

input_tokens = usage.prompt_token_count or 0
output_tokens = usage.candidates_token_count or 0
total_tokens = usage.total_token_count or 0

print("\nAPI Usage Summary:")
print("Input Tokens:", input_tokens)
print("Output Tokens:", output_tokens)
print("Total Tokens:", total_tokens)

# Basic usage check
MAX_TOTAL_TOKENS = 500

if total_tokens <= MAX_TOTAL_TOKENS:
    print("Token usage is within the configured limit.")
else:
    print("Token usage exceeded the configured limit.")
```

```python

# Practical 4 – Building an AI Chat UI with Gradio

import gradio as gr
from google import genai
from google.colab import userdata
from google.genai import types

# Initialize Gemini client
gemini_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=gemini_key)

# Define custom instructions
SYSTEM_INSTRUCTION = """
You are a friendly and helpful Python programming tutor.
Explain concepts in simple language.
Provide practical examples.
Keep answers beginner-friendly.
"""

# Handle chat messages
def chat_with_gemini(message, history):

    # Convert Gradio conversation history into Gemini messages
    contents = []

    for item in history:
        contents.append({
            "role": "user",
            "parts": [{"text": item["content"]}]
        }) if item["role"] == "user" else contents.append({
            "role": "model",
            "parts": [{"text": item["content"]}]
        })

    # Add current user message
    contents.append({
        "role": "user",
        "parts": [{"text": message}]
    })

    try:
        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=contents,
            config=types.GenerateContentConfig(
                system_instruction=SYSTEM_INSTRUCTION,
                temperature=0.5,
                max_output_tokens=500
            )
        )

        return response.text

    except Exception as e:
        return f"Sorry, an error occurred: {type(e).__name__}"

# Create Gradio interface
demo = gr.ChatInterface(
    fn=chat_with_gemini,
    title="My AI Python Tutor",
    description="Ask questions about Python programming!",
    textbox=gr.Textbox(
        placeholder="Ask me anything about Python...",
        container=False
    )
)

# Launch application
demo.launch(share=True, debug=False)
```
