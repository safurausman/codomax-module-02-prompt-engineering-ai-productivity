```python

# Day 6 - Practical 1: Zero-Shot Prompting

ticket = """
The application crashes whenever I try to save an order.
"""

prompt = f"""
Classify the following support ticket into exactly
one of these categories:
Bug, Feature Request, or General Inquiry.

Return only the category.

Ticket: {ticket}
"""

print("ZERO-SHOT PROMPT")
print(prompt)
```

```python

# Day 6 - Practical 3: Role-Based Prompting

prompt = """
Act as a senior .NET software architect.

Explain dependency injection in C#.

Requirements:
1. Explain the concept in simple language.
2. Provide a practical ASP.NET Core example.
3. Explain Singleton, Scoped, and Transient lifetimes.
4. Include common mistakes developers should avoid.
"""

print("ROLE-BASED PROMPT")
print(prompt)
```

```python

# Day 6 - Practical 4: Context, Instructions and Constraints

context = """
We are developing an ASP.NET Core Web API
using C# and Entity Framework Core.
"""

instructions = """
Act as a senior .NET developer.
Review the following method and suggest improvements.
Focus on readability, performance, and error handling.
"""

constraints = """
- Do not change the method signature.
- Do not introduce additional libraries.
- Keep the solution compatible with the existing project.
- Explain each suggested improvement.
"""

code = """
public async Task<Order> GetOrder(int id)
{
    var order = await _context.Orders.FindAsync(id);
    return order;
}
"""

prompt = f"""
CONTEXT:
{context}

INSTRUCTIONS:
{instructions}

CONSTRAINTS:
{constraints}

CODE:
{code}
"""

print(prompt)
```

Day 8 – AI-Assisted Coding and Debugging

```python

# Day 8 - Practical 1: AI-Assisted Code Generation

prompt = """
Act as a senior C# developer.

Generate a C# method named CalculateDiscount.

Requirements:
1. Accept the original price and discount percentage.
2. Calculate the discount amount.
3. Return the final price after discount.
4. Use decimal data types.
5. Validate that the price is not negative.
6. Validate that the discount percentage is between 0 and 100.
7. Include appropriate exception handling.
8. Provide a short explanation of the implementation.
"""

print("AI-ASSISTED CODE GENERATION PROMPT")
print(prompt)
```

```python

# Day 8 - Practical 2: AI-Assisted Debugging

buggy_code = """
public static bool IsEven(int number)
{
    return number % 2 == 1;
}
"""

prompt = f"""
Act as an experienced C# debugging specialist.

Review the following C# code.

Identify the logical error and explain why
the method produces incorrect results.

Requirements:
1. Explain the root cause.
2. Provide the corrected C# code.
3. Explain the correction.
4. Provide three test cases covering
   even and odd numbers.

Do not change the method name or signature.

Code:
{buggy_code}
"""

print("AI-ASSISTED DEBUGGING PROMPT")
print(prompt)
```

```python

# Day 8 - Practical 3: AI-Assisted Code Review

code = """
public decimal CalculateTotal(
    decimal price, int quantity)
{
    decimal total = 0;

    for (int i = 0; i < quantity; i++)
    {
        total = total + price;
    }

    return total;
}
"""

prompt = f"""
Act as a senior C# code reviewer.

Review the following C# method.

Requirements:
1. Identify potential performance issues.
2. Suggest improvements to readability.
3. Identify missing input validation.
4. Provide an improved implementation.
5. Explain the changes.

Keep the method name and signature unchanged.
Do not introduce additional libraries.

Code:
{code}
"""

print("AI-ASSISTED CODE REVIEW PROMPT")
print(prompt)
```

```python

# Day 8 - Practical 4: AI-Assisted Test Case Generation

code = """
public static decimal CalculateDiscount(
    decimal originalPrice,
    decimal discountPercentage)
{
    if (originalPrice < 0)
        throw new ArgumentException(
            "Price cannot be negative.");

    if (discountPercentage < 0 ||
        discountPercentage > 100)
        throw new ArgumentException(
            "Discount must be between 0 and 100.");

    decimal discountAmount =
        originalPrice * discountPercentage / 100;

    return originalPrice - discountAmount;
}
"""

prompt = f"""
Act as a senior C# software tester.

Generate test cases for the following method.

Requirements:
1. Include positive test cases.
2. Include negative test cases.
3. Include boundary-value test cases.
4. Specify input values and expected results.
5. Include a short testing report.

Use a clear table format.

Code:
{code}
"""

print("AI-ASSISTED TEST CASE GENERATION PROMPT")
print(prompt)
```

Day 9 – Reusable Prompt Templates


1. Reusable prompt templates reduce repetitive prompt writing.
2. Python functions help organize and reuse prompt structures.
3. Dynamic parameters allow prompts to accept different inputs.
4. Summarization templates support consistent content summaries.
5. Email templates help generate professional communications.
6. Coding templates help developers create structured coding instructions.
7. Reusable prompts improve consistency and productivity.

```python

# Day 9 - Practical 1: Reusable Prompt Template

def create_prompt(role, task, context, requirements):
    prompt = f"""
Role: {role}

Task: {task}

Context:
{context}

Requirements:
{requirements}
"""
    return prompt


prompt = create_prompt(
    role="Senior .NET Developer",
    task="Review the provided C# code",
    context="ASP.NET Core Web API using Entity Framework Core",
    requirements="""
    - Identify potential bugs.
    - Suggest performance improvements.
    - Improve readability.
    - Preserve existing functionality.
    """
)

print(prompt)
```

```python

# Day 9 - Practical 2: Summarization Template

def summarize_prompt(text, bullet_count=3):
    return f"""
Act as a professional summarizer.

Summarize the following text in {bullet_count} bullet points.

Requirements:
- Preserve the main ideas.
- Use simple language.
- Avoid adding unsupported information.

Text:
{text}
"""


article = """
Generative AI helps developers generate code,
debug applications, write documentation,
and improve software development productivity.
"""

prompt = summarize_prompt(article, 3)
print(prompt)
```

```python

# Day 9 - Practical 3: Email Writing Template

def email_prompt(recipient, purpose, tone):
    return f"""
Act as a professional business communication assistant.

Write an email to: {recipient}

Purpose:
{purpose}

Tone: {tone}

Requirements:
- Include a suitable subject.
- Keep the email concise.
- Use professional language.
- Include an appropriate closing.
"""


prompt = email_prompt(
    recipient="Project Manager",
    purpose="Request an update on project progress",
    tone="Professional and polite"
)

print(prompt)
```

```python

# Day 9 - Practical 4: Coding Assistance Template

def coding_prompt(language, task, constraints):
    return f"""
Act as an experienced {language} developer.

Task:
{task}

Constraints:
{constraints}

Requirements:
- Provide clean and readable code.
- Explain the implementation.
- Include error handling where appropriate.
- Follow the specified constraints.
"""


prompt = coding_prompt(
    language="C#",
    task="Create a method to retrieve an order by ID",
    constraints="""
    - Use async/await.
    - Use Entity Framework Core.
    - Do not introduce additional libraries.
    """
)

print(prompt)
```

Day 10 – Final Project: AI Productivity Toolkit


1. AI productivity workflows help simplify repetitive tasks.
2. Reusable prompts support summarization, email writing, coding, and research.
3. Python functions help organize and reuse prompt templates.
4. Testing verifies that toolkit functions generate valid prompts.
5. Human review and validation remain important when using AI-generated outputs.


```python

# Day 10 - AI Productivity Toolkit

def summarize_text(text):
    return f"""
Summarize the following text into 3 bullet points:
{text}
"""

def write_email(purpose, tone="Professional"):
    return f"""
Write a business email.
Purpose: {purpose}
Tone: {tone}
Include a subject and professional closing.
"""

def coding_assistant(task):
    return f"""
Act as a senior C# developer.
Task: {task}
Provide clean code, explanation, and error handling.
"""

def research_assistant(topic):
    return f"""
Act as a research assistant.
Research topic: {topic}
Provide introduction, findings, benefits,
challenges, and conclusion.
"""

print("AI PRODUCTIVITY TOOLKIT")
print("=======================")

print("\n1. SUMMARIZATION")
print(summarize_text("Generative AI improves developer productivity."))

print("\n2. EMAIL WRITING")
print(write_email("Request a project status update"))

print("\n3. CODING ASSISTANT")
print(coding_assistant("Create a C# method to retrieve an order by ID"))

print("\n4. RESEARCH ASSISTANT")
print(research_assistant("Generative AI in software development"))
```

```python

# Day 10 - Toolkit Testing

tests = {
    "Summarization": summarize_text("AI helps developers."),
    "Email Writing": write_email("Request a meeting"),
    "Coding Assistant": coding_assistant("Create a REST API"),
    "Research Assistant": research_assistant("Artificial Intelligence")
}

for name, result in tests.items():
    status = "PASS" if isinstance(result, str) and len(result.strip()) > 0 else "FAIL"
    print(f"{name}: {status}")
```
